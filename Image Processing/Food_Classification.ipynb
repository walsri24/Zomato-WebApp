```python
import modelbit
import base64
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
import numpy as np
from io import BytesIO
from PIL import Image

# Define your categories
category = {
    0: ['American', 'Baked Potato'], 1: ['American', 'Crispy Chicken'],
    2: ['Desserts', 'Donut'], 3: ['American', 'Fries'], 4: ['American', 'Hot Dog'],
    5: ['Fast Food', 'Sandwich'], 6: ['Mexican', 'Taco'], 7: ['Mexican', 'Taquito'],
    8: ['Desserts', 'Apple Pie'], 9: ['Burger', 'Burger'], 10: ['North Indian', 'Butter Naan'],
    11: ['Beverages', 'Chai'], 12: ['North Indian', 'Chapati'], 13: ['Bakery', 'Cheesecake'],
    14: ['North Indian', 'Chicken Curry'], 15: ['North Indian', 'Chole Bhature'],
    16: ['North Indian', 'Dal Makhani'], 17: ['Gujarati', 'Dhokla'],
    18: ['Asian', 'Fried Rice'], 19: ['Ice Cream', 'Ice Cream'],
    20: ['South Indian', 'Idli'], 21: ['Mithai', 'Jalebi'], 22: ['Street Food', 'Kaathi Rolls'],
    23: ['North Indian', 'Kadai Paneer'], 24: ['Desserts', 'Kulfi'],
    25: ['South Indian', 'Masala Dosa'], 26: ['Tibetan', 'Momos'], 27: ['American', 'Omelette'],
    28: ['Street Food', 'Paani Puri'], 29: ['Street Food', 'Pakode'], 30: ['Street Food', 'Pav Bhaji'],
    31: ['Italian', 'Pizza'], 32: ['Street Food', 'Samosa'], 33: ['Japanese', 'Sushi']
}



# Load your model globally
path_to_model = 'model_v1_inceptionV3.h5'
model = load_model(path_to_model)

# Prediction function
def predict_image(image_base64: str):
    # Decode the base64 image
    img_bytes = base64.b64decode(image_base64)
    img = Image.open(BytesIO(img_bytes))

    # Preprocess the image
    img = img.resize((299, 299))  # Resize to the required input size
    img_array = image.img_to_array(img)
    img_processed = np.expand_dims(img_array, axis=0)
    img_processed /= 255.

    # Make a prediction
    prediction = model.predict(img_processed)
    index = np.argmax(prediction)

    # Return the predicted category name
    return category[index][0], category[index][1]

# Deploy the function using Modelbit
modelbit.deploy(predict_image)
```

```python
import base64

def image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        # Read the image and encode it in base64
        encoded_string = base64.b64encode(image_file.read()).decode('utf-8')
    return encoded_string

# Example usage:
image_path = r'Zomato WebApp Backend\Image Processing\predict.ipynb'
image_base64 = image_to_base64(image_path)
```

```python
modelbit.get_inference(
  region="ap-south-1",
  workspace="prajwalswaroopsrivastava",
  deployment="predict_image",
  data=image_base64
)
```
