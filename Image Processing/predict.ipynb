```python
import modelbit
import base64
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
import numpy as np
from io import BytesIO
from PIL import Image

# Define your categories
category = {
    0: ['baked_potato', 'Baked Potato'], 1: ['crispy_chicken', 'Crispy Chicken'], 
    2: ['donut', 'Donut'], 3: ['fries', 'Fries'], 4: ['hot_dog', 'Hot Dog'], 
    5: ['sandwich', 'Sandwich'], 6: ['taco', 'Taco'], 7: ['taquito', 'Taquito'], 
    8: ['apple_pie', 'Apple Pie'], 9: ['burger', 'Burger'], 10: ['butter_naan', 'Butter Naan'], 
    11: ['chai', 'Chai'], 12: ['chapati', 'Chapati'], 13: ['cheesecake', 'Cheesecake'], 
    14: ['chicken_curry', 'Chicken Curry'], 15: ['chole_bhature', 'Chole Bhature'], 
    16: ['dal_makhani', 'Dal Makhani'], 17: ['dhokla', 'Dhokla'], 
    18: ['fried_rice', 'Fried Rice'], 19: ['ice_cream', 'Ice Cream'], 
    20: ['idli', 'Idli'], 21: ['jalebi', 'Jalebi'], 22: ['kaathi_rolls', 'Kaathi Rolls'], 
    23: ['kadai_paneer', 'Kadai Paneer'], 24: ['kulfi', 'Kulfi'], 
    25: ['masala_dosa', 'Masala Dosa'], 26: ['momos', 'Momos'], 27: ['omelette', 'Omelette'], 
    28: ['paani_puri', 'Paani Puri'], 29: ['pakode', 'Pakode'], 30: ['pav_bhaji', 'Pav Bhaji'], 
    31: ['pizza', 'Pizza'], 32: ['samosa', 'Samosa'], 33: ['sushi', 'Sushi']
}


# Load your model globally
path_to_model = 'E:\\AWORK\\PRAJWAL\\Coding\\iiit-kota-assign-walsri24\\Zomato WebApp Backend\\Image Processing\\model_v1_inceptionV3.h5'
model = load_model(path_to_model)

# Prediction function
def predict_image(image_base64: str):
    # Decode the base64 image
    img_bytes = base64.b64decode(image_base64)
    img = Image.open(BytesIO(img_bytes))
    
    # Preprocess the image
    img = img.resize((299, 299))  # Resize to the required input size
    img_array = image.img_to_array(img)
    img_processed = np.expand_dims(img_array, axis=0)
    img_processed /= 255.

    # Make a prediction
    prediction = model.predict(img_processed)
    index = np.argmax(prediction)
    
    # Return the predicted category name
    return category[index][1]

# Deploy the function using Modelbit
modelbit.deploy(predict_image)
```

```python
import base64

def image_to_base64(image_path):
    with open(image_path, "rb") as image_file:
        # Read the image and encode it in base64
        encoded_string = base64.b64encode(image_file.read()).decode('utf-8')
    return encoded_string

# Example usage:
image_path = r'E:\AWORK\PRAJWAL\Coding\iiit-kota-assign-walsri24\Zomato WebApp Backend\Image Processing\ice_cream.jpg'
image_base64 = image_to_base64(image_path)
```

```python
modelbit.get_inference(
  region="ap-south-1",
  workspace="prajwalswaroopsrivastava",
  deployment="predict_image",
  data=image_base64
)
```
